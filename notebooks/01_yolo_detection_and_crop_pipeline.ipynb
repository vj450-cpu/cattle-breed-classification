```python
input_folder = r"C:\Users\Vijay\Desktop\cattle"
output_folder = r"C:\Users\Vijay\Desktop\clean_dataset"

import os
import cv2
import shutil

os.makedirs(output_folder, exist_ok=True)

for breed in os.listdir(input_folder):
    breed_path = os.path.join(input_folder, breed)

    if not os.path.isdir(breed_path):
        continue

    save_breed_path = os.path.join(output_folder, breed)
    os.makedirs(save_breed_path, exist_ok=True)

    for file in os.listdir(breed_path):
        if file.lower().endswith((".jpg", ".jpeg", ".png")):

            img_path = os.path.join(breed_path, file)

            try:
                image = cv2.imread(img_path)
                results = model(img_path)

                detections = results[0].boxes

                cow_count = 0
                cow_box = None

                for box in detections:
                    if int(box.cls) == 19:  # cow class
                        cow_count += 1
                        cow_box = box

                # keep only SINGLE cow
                if cow_count == 1:
                    x1, y1, x2, y2 = map(int, cow_box.xyxy[0])
                    cropped = image[y1:y2, x1:x2]

                    save_path = os.path.join(save_breed_path, file)
                    cv2.imwrite(save_path, cropped)

            except:
                pass

print("✅ Dataset cleaned!")
```

```python
from sklearn.model_selection import train_test_split
import os
import shutil

base = r"C:\Users\Vijay\Desktop\clean_dataset"
train_dir = r"C:\Users\Vijay\Desktop\data\train"
test_dir = r"C:\Users\Vijay\Desktop\data\test"

for breed in os.listdir(base):
    breed_path = os.path.join(base, breed)

    if not os.path.isdir(breed_path):
        continue

    imgs = os.listdir(breed_path)

    train_imgs, test_imgs = train_test_split(imgs, test_size=0.2)

    os.makedirs(os.path.join(train_dir, breed), exist_ok=True)
    os.makedirs(os.path.join(test_dir, breed), exist_ok=True)

    for img in train_imgs:
        shutil.copy(
            os.path.join(base, breed, img),
            os.path.join(train_dir, breed, img)
        )

    for img in test_imgs:
        shutil.copy(
            os.path.join(base, breed, img),
            os.path.join(test_dir, breed, img)
        )

print("✅ Train/Test split done!")
```

```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import timm

# 🔥 Device (VERY IMPORTANT)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# 🟢 Train Transform (with augmentation)
train_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(0.3,0.3,0.3),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

# 🔵 Test Transform (NO augmentation ❗)
test_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

# 📂 Dataset
train_data = datasets.ImageFolder("data/train", transform=train_transform)
test_data = datasets.ImageFolder("data/test", transform=test_transform)

train_loader = DataLoader(train_data, batch_size=16, shuffle=True)
test_loader = DataLoader(test_data, batch_size=16)

# 🧠 Model
model = timm.create_model('vit_base_patch16_224', pretrained=True)
model.head = nn.Linear(model.head.in_features, len(train_data.classes))
model = model.to(device)

# ⚙️ Training setup
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.00003)

# 🚀 Training loop
for epoch in range(20):
    model.train()
    running_loss = 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        outputs = model(images)
        loss = criterion(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {running_loss:.4f}")

# ✅ Evaluation
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)

        outputs = model(images)
        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print(f"🔥 Test Accuracy: {accuracy:.2f}%")

# 💾 Save model
torch.save(model.state_dict(), "vit_model.pth")
```
